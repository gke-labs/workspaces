# NumPy and Matplotlib Example

A simple notebook demonstrating data generation with NumPy and visualization with Matplotlib.

## Import Required Libraries

Import NumPy for numerical computation and Matplotlib for plotting.

```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

## Generate Data with NumPy

Create a sine wave and some random data to visualize.

```python
rng = np.random.default_rng(seed=42)

x = np.linspace(0, 2 * np.pi, 200)
y_sine = np.sin(x)

scatter_x = rng.uniform(0, 10, 100)
scatter_y = 2 * scatter_x + rng.normal(0, 3, 100)

hist_data = rng.normal(loc=0, scale=1, size=1000)
```

## Line Plot: Sine Wave

Plot the sine wave generated above.

```python
plt.figure(figsize=(8, 4))
plt.plot(x, y_sine, color="tab:blue", label="sin(x)")
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.grid(True)
plt.show()
```

## Scatter Plot: Random Data

Plot noisy linear data as a scatter plot.

```python
plt.figure(figsize=(8, 4))
plt.scatter(scatter_x, scatter_y, color="tab:orange", alpha=0.7)
plt.title("Noisy Linear Data")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

## Histogram: Normal Distribution

Plot a histogram of normally distributed random values.

```python
plt.figure(figsize=(8, 4))
plt.hist(hist_data, bins=30, color="tab:green", edgecolor="black", alpha=0.7)
plt.title("Histogram of Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```
